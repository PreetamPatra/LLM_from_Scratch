```python
import tiktoken
```

```python
tokenizer=tiktoken.get_encoding("cl100k_base")
```

```python
text="I am an LLM.|<endoftext>| I was created by Preetam"
```

```python
tokens=tokenizer.encode(text,allowed_special={"<|endoftext|>"})
print(tokens)
```

```python
characters=tokenizer.decode(tokens)
print(characters)
```
